```python
from pystac_client import Client
```

```python
api_url = "https://earth-search.aws.element84.com/v1/"
```

```python
client = Client.open(api_url)
```

```python
client
```

```python
collections = client.get_collections()
```

```python
collections = list(collections)
```

```python
for collection in collections:
    print(collection)
```

```python
s2_collection = client.get_collection('sentinel-2-l2a')
```

```python
s2_collection
```

```python
poi = {"type": "Point", "coordinates": (6.854188, 52.237955)}
```

```python
search_results = client.search(
    collections=["sentinel-2-l2a"],
    intersects=poi,
    datetime="2023-07",
    max_items=1,
)
```

```python
print(search_results.matched())
```

```python
items = search_results.item_collection()
items
```

```python
item = items[0]
```

```python
item
```

```python
url_address = item.assets["red"].href
url_address
```

```python
import requests
```

```python
with open("red.tif", "wb") as file:
    file.write(requests.get(url_address).content)
```
