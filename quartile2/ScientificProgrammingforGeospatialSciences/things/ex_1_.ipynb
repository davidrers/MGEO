```python
def calculate_schawarzschild_radius(mass_kg):
    """
    Calculate the Schwarzschild radius of a non-rotating black hole.

    Parameters:
    mass_kg (float): Mass of the black hole in kilograms.

    Returns:
    float: Schwarzschild radius in meters.
    """
    G = 6.67430e-11  # Gravitational constant in m^3 kg^-1 s^-2
    c = 299792458    # Speed of light in m/s

    radius = (2 * G * mass_kg) / (c ** 2)
    return radius
```
