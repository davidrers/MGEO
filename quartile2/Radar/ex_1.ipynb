```python
import scipy.io
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.pyplot import figure

mat = scipy.io.loadmat('C:\\Users\\darey\\Downloads\\rangeline.mat')
```

```python

s = mat['srl04986'].flatten()
real_part = np.real(s)
imag_part = np.imag(s)

plt.figure()
plt.plot(real_part)
plt.title("Real part of SAR range line")
plt.xlabel("Sample index")
plt.ylabel("Real value")
plt.grid(True)

plt.figure()
plt.plot(imag_part)
plt.title("Imaginary part of SAR range line")
plt.xlabel("Sample index")
plt.ylabel("Imaginary value")
plt.grid(True)

print("Real part: min =", real_part.min(), "max =", real_part.max())
print("Imag part: min =", imag_part.min(), "max =", imag_part.max())
```
