### Step 4: Connect to the Database

Set your MongoDB URI and database name. Defaults target a local server on `mongodb://localhost:27017` and a database named `nyc_spatial_data`. If your instructor gave you a different URI/credentials, set `MONGODB_URI` and `MONGODB_DB` as environment variables (or edit the variables below).

```python
# Importing the 'pymongo' module for MongoDB interaction
import os
import pymongo

database = None

# Allow overriding via environment variables
MONGODB_URI = os.getenv("MONGODB_URI", "mongodb://localhost:27017/")
DB_NAME = os.getenv("MONGODB_DB", "nyc_spatial_data")

# Try to create a MongoDB client and connect to the server
try:
    client = pymongo.MongoClient(MONGODB_URI)
    database = client[DB_NAME]

    print(f"Connected to MongoDB at {MONGODB_URI}")
    print(f"Using database: {DB_NAME}")
    print("Collections:", database.list_collection_names())
except Exception as e:
    print(f"Error: {e}")
```

### Step 5: Create a Collection

Create (or access) a `students` collection. MongoDB creates the collection on first insert if it does not already exist.

```python
def create_student_collection():
    try:
        # Creating a collection named 'students' within the selected database
        collection = database["students"]

        # Printing a message indicating a successful collection creation
        print("Collection 'students' created successfully")
        return collection

    except Exception as e:
        # Handling exceptions and printing an error message if collection creation fails
        print(f"Error: {e}")

my_col = create_student_collection()
```

### Step 6: Insert a Single Document

Insert one student document and view the inserted ID.

```python
mydict = { "name": "John", "address": "Highway 37" }

x = my_col.insert_one(mydict)

print(x.inserted_id)
```

### Step 7: Find a Single Document

Retrieve one document to verify the insert worked.

```python
x = my_col.find_one()

print(x)
```

### Step 8: Insert Many Documents

Bulk insert multiple documents into the same collection.

```python
mylist = [
  { "name": "Amy", "address": "Apple st 652"},
  { "name": "Hannah", "address": "Mountain 21"},
  { "name": "Michael", "address": "Valley 345"},
  { "name": "Sandy", "address": "Ocean blvd 2"},
  { "name": "Betty", "address": "Green Grass 1"},
  { "name": "Richard", "address": "Sky st 331"},
  { "name": "Susan", "address": "One way 98"},
  { "name": "Vicky", "address": "Yellow Garden 2"},
  { "name": "Ben", "address": "Park Lane 38"},
  { "name": "William", "address": "Central st 954"},
  { "name": "Chuck", "address": "Main Road 989"},
  { "name": "Viola", "address": "Sideway 1633"}
]

x = my_col.insert_many(mylist)

#print list of the _id values of the inserted documents:
print(x.inserted_ids)
```
