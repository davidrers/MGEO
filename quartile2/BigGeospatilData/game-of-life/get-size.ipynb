```python
from pympler import asizeof
```

```python
import random

def create_grid(rows, cols, alive):
    grid = [[0 for x in range(cols)] for y in range(rows)]
    count = int(rows * cols * alive)
    while count:
        row, col = random.randrange(rows), random.randrange(cols)
        if grid[row][col] == 0:
            grid[row][col] = 1
            count -= 1
    return grid
```

```python
def create_matrix(rows, cols):
    grid = []
    val, delta = 0.0, 1e-10
    for _ in range(rows):
        row = []
        for _ in range(cols):
            row.append(val)
            val += delta
        grid.append(row)
    return grid
```

```python
import sys

def get_size(grid):
    size = sys.getsizeof(grid)
    for row in grid:
        size += sys.getsizeof(row)
        for cell in row:
            size += sys.getsizeof(cell)
    return size
```

```python
matrix = create_matrix(10_000, 10_000)
```

```python
size = get_size(matrix)
size
```

```python
size = asizeof.asizeof(matrix)
size
```

```python
grid = create_grid(10_000, 10_000, 0.5)
```

```python
size = get_size(grid)
size
```

```python
size = asizeof.asizeof(grid)
size
```
