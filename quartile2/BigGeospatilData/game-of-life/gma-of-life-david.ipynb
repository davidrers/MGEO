```python
import random
%load_ext line_profiler
def create_grid(rows, cols, alive):
    grid = [[0]*cols for y in range(rows)]
    count = int(rows * cols * alive)
    while count:
        row, col = random.randrange(rows), random.randrange(cols)
        if grid[row][col] == 0:
            grid[row][col] = 1
            count -= 1
    return grid


def next_generation(grid):

    temp_grid=[[0]+i+[0] for i in [[0]*len(grid[0])]+grid+[[0]*len(grid[0])]]

    update_live=[]
    update_death=[]
    for row in range(1,len(temp_grid)-1):
        for col in range(1,len(temp_grid[0])-1):

            #ncount=sum (temp_grid[row+ dr][col + dc] for dr in (-1, 0, 1) for dc in (-1, 0, 1) if not (dr == 0 and dc == 0))

            ncount = (
                    temp_grid[row-1][col-1] + temp_grid[row-1][col]   + temp_grid[row-1][col+1] +
                    temp_grid[row][col-1]   +                   temp_grid[row][col+1]   +
                    temp_grid[row+1][col-1] + temp_grid[row+1][col]   + temp_grid[row+1][col+1]
                )

            if grid[row-1][col-1]==1:
                if ncount <2:
                    update_death.append((row-1,col-1))
                if ncount in [2,3]:
                    pass
                if ncount >3:
                    update_death.append((row-1,col-1))
            elif grid[row-1][col-1]==0:
                if ncount==3:
                    update_live.append((row-1,col-1))

    new_grid = [row[:] for row in grid]
    for cell in update_live:
        new_grid[cell[0]][cell[1]]=1

    for cell in update_death:
        new_grid[cell[0]][cell[1]]=0
            
    return new_grid

grid=create_grid(500,500,0.5)
```

```python
%%timeit -n 5 -r 3
next_grid=next_generation(grid)
```

```python
%lprun -f next_generation next_generation(grid)
```
